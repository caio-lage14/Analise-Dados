```python
import pandas as pd

caminho_arquivo = r"H:\python\3-etapa\atividade_4\CD2022_Populacao_2010_Compatibilizada_20231222 (1).xlsx"

df = pd.read_excel(caminho_arquivo, sheet_name="Municípios", header=2)

df = df.drop(columns=[c for c in df.columns if "Unnamed" in str(c)])
df = df.dropna(subset=["UF"]).reset_index(drop=True)

df = df.rename(columns={
    "População Município 2010\n(Sinopse)": "Pop_2010_Sinopse",
    "População 2010 (Alterações de Limites até 2022)1": "Pop_2010_Compatibilizada",
    "População Censo 2022": "Pop_2022"
})

print(df.head())
```

```python
df = df.iloc[0:5570]
```

```python
df["Dif"] = df["Pop_2022"] - df["Pop_2010_Compatibilizada"]
```

```python
df
```

```python
pop_estado = (
    df.groupby("UF")[["Pop_2010_Compatibilizada", "Pop_2022", "Dif"]]
    .sum()
    .reset_index()
)
pop_estado

```

```python
pop_estado.to_csv(r"H:\python\3-etapa\atividade_4\pop_estado.csv", sep =";")
```
