```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
path2026 =r"H:\Python\3EtapaPY\Atv_6_MATPLOTLIB_MELT_MERGE\Tabela5-sem_emprego_2026.csv"
path2012 =r"H:\Python\3EtapaPY\Atv_6_MATPLOTLIB_MELT_MERGE\Tabela5-sem_emprego_2012.csv"
path2 =r"H:\Python\3EtapaPY\Atv_6_MATPLOTLIB_MELT_MERGE\Tabela 1.1.1.xls"
```

```python

tab2 = pd.read_excel(path2, engine="xlrd", header=None)
tab2012 = pd.read_csv(path2012, sep = ";", encoding = "utf8", decimal=",")
tab2026 = pd.read_csv(path2026, sep = ";", encoding = "utf8", decimal=",")
```

```python
tab2012 = tab2012.rename(columns={"Desocupados - homens (2012 T1)": "Homens 2012"})
tab2012 = tab2012.rename(columns={"Desocupados - mulheres (2012 T1)": "Mulheres 2012"})
tab2012
```

```python
tab2026 = tab2026.rename(columns={"Desocupados - homens (2026 T1)": "Homens 2026"})
tab2026 = tab2026.rename(columns={"Desocupados - mulheres (2026 T1)": "Mulheres 2026"})
tab2026
```

```python
comp = tab2012.merge(tab2026,on=["Sigla", "Estado", "Código"], how= "inner")
```

```python
comp
```

```python
comp.columns
```

```python
ordem = comp.sort_values("Mulheres 2026")["Estado"]
longo = comp.melt(
    id_vars= ["Sigla", "Código", "Estado"],
    value_vars=["Mulheres 2026", "Mulheres 2012"],
    var_name="Ano",
    value_name="Participacao_mulheres",
)
longo["Ano"] = longo["Ano"].map({"Mulheres 2012": "2012 T1", "Mulheres 2026": "2026 T1"})
```

```python
fig, ax = plt.subplots(figsize=(10, 10))
sns.barplot(
    data=longo,
    y="Estado",
    x="Participacao_mulheres",
    hue="Ano",
    order=ordem,
    ax=ax,
)
ax.axvline(50, color="Green", linestyle="--", linewidth=1) 
ax.set_xlabel("Participação das mulheres entre as pessoas desocupadas (%)")
ax.set_ylabel("")
ax.set_title("Desocupação: participação feminina em 2012 T1 e 2026 T1")
ax.legend(title="Taxa de Desemprego")
fig.tight_layout()
plt.show()
```
